## Q7. Sum of weights

Compute $w = (X^T X)^{-1} X^T y$ for the first 7 cars from Asia, using `vehicle_weight` and `model_year`.

````python
X = df[df['origin']=='Asia'][['vehicle_weight','model_year']].head(n=7).values
````

````python
XTX = X.T.dot(X)
XTX
````

````python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
````

````python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
````

````python
w = XTX_inv.dot(X.T).dot(y)
````

````python
w.sum()
````

**Answer:** 0.369
